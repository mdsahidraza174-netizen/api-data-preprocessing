```python
data = response.json()
```

```python
rows = []

for match in data:
    for bookmaker in match['bookmakers']:
        rows.append({
            "home_team": match['home_team'],
            "away_team": match['away_team'],
            "bookmaker": bookmaker['title'],
            "last_update": bookmaker['last_update']
        })
```

```python
df = pd.DataFrame(rows)

```
